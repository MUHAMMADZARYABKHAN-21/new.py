```python
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

# Create dataset
X, y = make_moons(
    n_samples=300,
    noise=0.08,
    random_state=42
)

# Different eps values
eps_values = [0.05, 0.10, 0.20, 0.50]

for eps in eps_values:

    model = DBSCAN(
        eps=eps,
        min_samples=5
    )

    labels = model.fit_predict(X)

    # Number of clusters (ignore -1 because -1 = noise)
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

    # Number of noise points
    n_noise = list(labels).count(-1)

    print(
        f"eps = {eps} | "
        f"clusters = {n_clusters} | "
        f"noise = {n_noise}"
    )

    # Plot
    plt.figure(figsize=(6, 4))
    plt.scatter(X[:, 0], X[:, 1], c=labels)

    plt.title(f"DBSCAN — eps = {eps}")
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")

    plt.show()
```

```python
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

X, y = make_moons(
    n_samples=300,
    noise=0.08,
    random_state=42
)
```

```python
min_samples_values = [2, 5, 10, 20]

for m in min_samples_values:

    model = DBSCAN(
        eps=0.2,
        min_samples=m
    )

    labels = model.fit_predict(X)

    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    n_noise = list(labels).count(-1)

    print(
        f"min_samples = {m} | "
        f"clusters = {n_clusters} | "
        f"noise = {n_noise}"
    )

    plt.figure(figsize=(6, 4))
    plt.scatter(X[:, 0], X[:, 1], c=labels)

    plt.title(f"DBSCAN — min_samples = {m}")
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.show()
```
