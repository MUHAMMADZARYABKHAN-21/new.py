```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

iris = load_iris()
X = iris.data
y = iris.target  # ✅ Correct: use target labels, not data
print("Original data shape:", X.shape)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("PCA shape:", X_pca.shape)

plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=y,  # class labels
    cmap='plasma',  # map labels to colors
    s=40,  # point size
    edgecolor='k'  # black edges for clarity
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Iris Dataset After PCA")
plt.show()
```

```python
print(pca.explained_variance_ratio_)
```

```python
print(pca.explained_variance_ratio_.sum())
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

variance = pca.explained_variance_ratio_

print("PC1:", variance[0])
print("PC2:", variance[1])

print("Total:", variance.sum())
```

```python
print("PC1:", variance[0] * 100, "%")
print("PC2:", variance[1] * 100, "%")
print("Total:", variance.sum() * 100, "%")
```

```python
pca = PCA()

X_pca = pca.fit_transform(X_scaled)

cumulative_variance = pca.explained_variance_ratio_.cumsum()

print(cumulative_variance)
```
