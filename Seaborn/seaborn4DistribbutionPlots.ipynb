****When Should You Use a KDE Plot?

Use a KDE plot when you want to:

Understand the overall shape of your data.
Compare distributions between groups.
See where values are concentrated.
Avoid the blocky appearance of a histogram.

Avoid using KDE when:

Your dataset is very small (the curve may be misleading).
Your variable is categorical instead of numerical.****

```python
# ============================================
# Seaborn Lesson 3 - Distribution Plots
# histplot(), kdeplot(), displot()
# Dataset: Titanic1.csv
# ============================================

# Import Libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("Titanic1.csv")

# Preview Dataset
print(df.head())
print(df.info())

# ============================================
# 1. Basic Histogram
# ============================================

plt.figure(figsize=(8,5))
sns.histplot(data=df, x="age")
plt.title("Histogram of Age")
plt.show()

# ============================================
# 2. Histogram with KDE
# ============================================

plt.figure(figsize=(8,5))
sns.histplot(data=df, x="age", kde=True)
plt.title("Histogram with KDE")
plt.show()

# ============================================
# 3. Histogram with Custom Bins
# ============================================

plt.figure(figsize=(8,5))
sns.histplot(data=df, x="age", bins=20)
plt.title("Histogram with 20 Bins")
plt.show()

# ============================================
# 4. Histogram with Color
# ============================================

plt.figure(figsize=(8,5))
sns.histplot(data=df, x="age", color="green")
plt.title("Green Histogram")
plt.show()

# ============================================
# 5. Histogram by Gender
# ============================================

plt.figure(figsize=(8,5))
sns.histplot(data=df, x="age", hue="sex", kde=True)
plt.title("Age Distribution by Gender")
plt.show()

# ============================================
# 6. KDE Plot
# ============================================

plt.figure(figsize=(8,5))
sns.kdeplot(data=df, x="age")
plt.title("KDE Plot")
plt.show()

# ============================================
# 7. Filled KDE
# ============================================

plt.figure(figsize=(8,5))
sns.kdeplot(data=df, x="age", fill=True)
plt.title("Filled KDE")
plt.show()

# ============================================
# 8. Filled KDE with Green Color
# ============================================

plt.figure(figsize=(8,5))
sns.kdeplot(data=df, x="age", fill=True, color="green")
plt.title("Green Filled KDE")
plt.show()

# ============================================
# 9. KDE by Gender
# ============================================

plt.figure(figsize=(8,5))
sns.kdeplot(data=df, x="age", hue="sex", fill=True)
plt.title("Age Distribution by Gender")
plt.show()

# ============================================
# 10. KDE with Thick Line
# ============================================

plt.figure(figsize=(8,5))
sns.kdeplot(data=df, x="age", linewidth=3)
plt.title("Thick KDE Line")
plt.show()

# ============================================
# 11. KDE with Dashed Line
# ============================================

plt.figure(figsize=(8,5))
sns.kdeplot(data=df, x="age", linestyle="--")
plt.title("Dashed KDE")
plt.show()

# ============================================
# 12. displot - Histogram
# ============================================

sns.displot(data=df, x="age")
plt.show()

# ============================================
# 13. displot - KDE
# ============================================

sns.displot(data=df, x="age", kind="kde")
plt.show()

# ============================================
# 14. displot - Histogram + KDE
# ============================================

sns.displot(data=df, x="age", kde=True)
plt.show()

# ============================================
# 15. displot with Custom Bins
# ============================================

sns.displot(data=df, x="age", bins=20)
plt.show()

# ============================================
# 16. displot with Green Color
# ============================================

sns.displot(data=df, x="age", color="green")
plt.show()

# ============================================
# 17. displot by Gender (Separate Columns)
# ============================================

sns.displot(data=df, x="age", col="sex")
plt.show()

# ============================================
# 18. displot by Passenger Class
# ============================================

sns.displot(data=df, x="fare", col="pclass")
plt.show()

# ============================================
# 19. displot with hue
# ============================================

sns.displot(data=df, x="age", hue="sex", kde=True)
plt.show()

# ============================================
# 20. displot with Row and Column
# ============================================

sns.displot(
    data=df,
    x="fare",
    row="sex",
    col="pclass"
)
plt.show()

# ============================================
# 21. displot with Height and Aspect
# ============================================

sns.displot(
    data=df,
    x="age",
    height=5,
    aspect=1.5
)
plt.show()

print("All Distribution Plot Examples Completed Successfully!")
```
