## What is a Bar Plot?

A bar plot compares a numerical variable across categories.

Example:

Category → Passenger Class

Value → Average Fare

Seaborn calculates the mean by default.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
df=pd.read_csv("Titanic1.csv")
sns.barplot(
    data=df,
    x="pclass",
    y="fare"
)

plt.title("Average Fare by Passenger Class")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("Titanic1.csv")

# Barplot of average fare by passenger class
sns.barplot(
    data=df,
    x="pclass",
    y="fare",
    color="white",   # bar color
    edgecolor="black"  # optional: outline bars
)

# Set background color
plt.gca().set_facecolor("pink")

plt.title("Average Fare by Passenger Class")
plt.show()
```

- ## What is a Count Plot?

A Count Plot displays **how many observations belong to each category**.

It does **NOT** calculate averages, sums, or percentages.

It simply counts rows.

Examples:

Sex

Male      ███████████████
Female    ███████

Pclass

1 ██████
2 █████████
3 ███████████████

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("Titanic1.csv")
sns.countplot(
    data=df,
    x="sex"
)

plt.title("Number of Male and Female Passengers")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("Titanic1.csv")
sns.countplot(
    data=df,
    x="pclass",
    hue="sex",
    palette="Set2"
)

plt.title("Number of Male and Female Passengers")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("Titanic1.csv")
sns.countplot(
    data=df,
    y="sex",
    hue="sex",
    palette="Set2"
)

plt.title("Number of Male and Female Passengers")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("Titanic1.csv")
sns.countplot(
    data=df,
    y="sex",
    hue="survived",
    palette="Set2",

    order=["male", "female"]
)

plt.title("Number of Male and Female Passengers")
plt.show()
```

## What is a Box Plot?

A Box Plot summarizes the distribution of numerical data.

Instead of showing every data point, it shows:

- Minimum
- First Quartile (Q1)
- Median (Q2)
- Third Quartile (Q3)
- Maximum

It also highlights outliers.

Box plots are one of the most important tools in
Exploratory Data Analysis (EDA).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Titanic1.csv")
sns.boxplot(
    data=df,
    x="fare"
)
plt.title("Distribution of fare ")
plt.gca().set_facecolor("white")

plt.show()
```

Machine Learning Connection

Before training a model, data scientists often create box plots to detect outliers.

For example:

A passenger fare of £512 on the Titanic is much higher than most fares.
Such extreme values can influence some machine learning algorithms, especially Linear Regression.

Using a box plot, you can quickly identify these unusual values and decide whether to keep them, transform them, or remove them.
