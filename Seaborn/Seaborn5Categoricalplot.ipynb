## What is a Bar Plot?

A bar plot compares a numerical variable across categories.

Example:

Category → Passenger Class

Value → Average Fare

Seaborn calculates the mean by default.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
df=pd.read_csv("Titanic1.csv")
sns.barplot(
    data=df,
    x="pclass",
    y="fare"
)

plt.title("Average Fare by Passenger Class")

plt.show()
```

- ## What is a Count Plot?

A Count Plot displays **how many observations belong to each category**.

It does **NOT** calculate averages, sums, or percentages.

It simply counts rows.

Examples:

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("Titanic1.csv")
sns.countplot(
    data=df,
    x="sex"
)

plt.title("Number of Male and Female Passengers")
plt.show()
```

## What is a Box Plot?

A Box Plot summarizes the distribution of numerical data.

Instead of showing every data point, it shows:

- Minimum
- First Quartile (Q1)
- Median (Q2)
- Third Quartile (Q3)
- Maximum

It also highlights outliers.

Box plots are one of the most important tools in
Exploratory Data Analysis (EDA).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Titanic1.csv")
sns.boxplot(
    data=df,
    x="fare"
)
plt.title("Distribution of fare ")
plt.gca().set_facecolor("white")

plt.show()
```

Machine Learning Connection

Before training a model, data scientists often create box plots to detect outliers.

For example:

A passenger fare of £512 on the Titanic is much higher than most fares.
Such extreme values can influence some machine learning algorithms, especially Linear Regression.

Using a box plot, you can quickly identify these unusual values and decide whether to keep them, transform them, or remove them.

- ## What is a Strip Plot?

A Strip Plot displays **every individual observation** in the dataset.

Unlike a box plot, which summarizes the data, a strip plot shows every data point.

Each dot represents one row (one passenger in the Titanic dataset).

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
df=pd.read_csv("Titanic1.csv")
sns.stripplot(
    data=df,
    x="pclass",
    y="fare",
    hue="sex"

)
plt.title("Distribution of fare")
plt.show()


```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
df=pd.read_csv("Titanic1.csv")
plt.figure(figsize=(8,5))

sns.stripplot(
    data=df,
    x="pclass",
    y="fare",
    jitter=True
)

plt.show()
plt.title("Distribution of fare")
plt.show()


```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
df=pd.read_csv("Titanic1.csv")
plt.figure(figsize=(8,5))

plt.figure(figsize=(8,5))
sns.boxplot(
    data=df,
    x="pclass",
    y="fare",
    color="lightgray"
)

sns.stripplot(
    data=df,
    x="pclass",
    y="fare",
    jitter=True
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
df=pd.read_csv("Titanic1.csv")
plt.figure(figsize=(10,6))

sns.stripplot(
    data=df,
    x="embarked",
    y="fare",
    hue="sex",
    palette="Set2",
    jitter=True,
    size=6,
    alpha=0.7
)

plt.title("Passenger Fare by Port and Gender")
plt.xlabel("Port of Embarkation")
plt.ylabel("Fare")

plt.grid(axis="y")

plt.show()
```

## What is a Swarm Plot?

A Swarm Plot displays every individual data point, just like a Strip Plot.

However, instead of randomly spreading points (jitter),
Seaborn intelligently arranges them so they **never overlap**.

This gives a much clearer picture of the distribution.

- ## What is a Point Plot?

A Point Plot displays the **average** (or another estimator)
of a numerical variable for each category.

Unlike a bar plot, it **doesn't use bars**.

Instead, it connects the averages with lines.

It is useful for comparing trends between categories.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
df=pd.read_csv("Titanic1.csv")
plt.figure(figsize=(8,5))


sns.pointplot(
    data=df,
    x="pclass",
    y="fare"
)

plt.title("Average Fare by Passenger Class")

plt.show()



```

```python



plt.figure(figsize=(8,5))

sns.pointplot(
    data=df,
    x="pclass",
    y="fare",
    hue="sex"
)

plt.title("Average Fare by Passenger Class")

plt.show()

plt.figure(figsize=(8,5))

sns.pointplot(
    data=df,
    x="pclass",
    y="fare",
    marker="D"
)
plt.figure(figsize=(8,5))

sns.pointplot(
    data=df,
    x="pclass",
    y="fare",
    linestyle="--"
)

plt.show()
plt.figure(figsize=(8,5))

sns.pointplot(
    data=df,
    x="pclass",
    y="fare",
    errorbar=None
)

plt.show()
import numpy as np

plt.figure(figsize=(8,5))

sns.pointplot(
    data=df,
    x="pclass",
    y="fare",
    estimator=np.median,
    errorbar=None
)

plt.show()
plt.figure(figsize=(10,6))

sns.pointplot(
    data=df,
    x="embarked",
    y="fare",
    hue="sex",
    palette="Set2",
    marker="o",
    linestyle="-",
    errorbar=None
)

plt.title("Average Fare by Port and Gender")
plt.xlabel("Port of Embarkation")
plt.ylabel("Average Fare")

plt.grid(axis="y")

plt.show()

```
