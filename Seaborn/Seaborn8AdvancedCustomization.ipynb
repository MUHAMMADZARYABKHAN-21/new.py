```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Titanic1.csv")
sns.set_theme(
    style="whitegrid",
    context="notebook"
)

sns.histplot(data=df, x="age")

plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.xticks(rotation=90)
sns.despine()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Titanic1.csv")

sns.set_theme(
    style="whitegrid",
    context="notebook"
)

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="age",
    bins=20
)

plt.title(
    "Distribution of Passenger Age",
    fontsize=18,
    pad=15
)

plt.xlabel(
    "Passenger Age",
    fontsize=13
)

plt.ylabel(
    "Number of Passengers",
    fontsize=13
)

sns.despine()

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Titanic1.csv")

sns.set_theme(style="whitegrid")

sns.countplot(
    data=df,
    x="pclass",
    hue="sex",
    palette="Set2"
)

plt.title("Passengers by Class and Gender")
plt.show()
# "deep"
# "muted"
# "bright"
# "pastel"
# "dark"
# "colorblind"
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Titanic1.csv")

average_fare = df["fare"].mean()

sns.histplot(data=df, x="fare")

plt.axvline(
    average_fare,
    linestyle="--"
)

plt.text(
    average_fare,
    100,
    f"Mean Fare: {average_fare:.2f}"
)

plt.title("Passenger Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
average_age = df["age"].mean()

sns.histplot(data=df, y="age")

plt.axhline(
    average_age,
    linestyle="--"
)

plt.show()

plt.show()
```
