```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd



df=pd.read_csv("Titanic1.csv")

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived",
    style="sex",
    size="fare"
)
sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="sex",
    palette="coolwarm"
)

plt.show()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("Titanic1.csv")

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived",
    style="sex",
    size="fare",
    palette="viridis"
)

plt.title("Titanic Passengers")

plt.show()
```
