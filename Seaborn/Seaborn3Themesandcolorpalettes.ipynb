```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")
data=pd.read_csv("Titanic1.csv")
sns.swarmplot(
    data=data,
    x="sex",
    y="fare",
    hue="survived",

)
plt.show()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
data = pd.read_csv("Titanic1.csv")

sns.stripplot(
    data=data,
    x="sex",
    y="fare",
    hue="survived",
    dodge=True,   # separates points by hue
    jitter=True   # spreads points for visibility
)

plt.show()
```
