```python
Z_single = linkage(
    X_scaled,
    method="single"
)

plt.figure(figsize=(15, 7))

dendrogram(Z_single)

plt.title("Single Linkage - Iris Dataset")

plt.show()
```

```python
Z_complete = linkage(
    X_scaled,
    method="complete"
)

plt.figure(figsize=(15, 7))

dendrogram(Z_complete)

plt.title("Complete Linkage - Iris Dataset")

plt.show()
```

```python
Z_average = linkage(
    X_scaled,
    method="average"
)

plt.figure(figsize=(15, 7))

dendrogram(Z_average)

plt.title("Average Linkage - Iris Dataset")

plt.show()
```

```python
Z_ward = linkage(
    X_scaled,
    method="ward"
)

plt.figure(figsize=(15, 7))

dendrogram(Z_ward)

plt.title("Ward Linkage - Iris Dataset")

plt.show()
```
