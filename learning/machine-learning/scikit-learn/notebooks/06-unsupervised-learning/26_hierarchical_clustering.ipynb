6. Step 1 — Calculate Distances

Hierarchical clustering needs a distance measure.

We'll start with Euclidean distance.

For two points:

$$ A=(x_1,y_1) $$

and:

$$ B=(x_2,y_2) $$
he distance is:

$$ \boxed{ d(A,B)= \sqrt{(x_1-x_2)^2+(y_1-y_2)^2} } $$

For example:

$$ A=(1,20) $$ $$ B=(2,25) $$

Then:

$$ d(A,B) = \sqrt{(1-2)^2+(20-25)^2} $$ $$ = \sqrt{1+25} $$ $$ =\sqrt{26} $$

10. Linkage

Linkage defines:

How do we measure the distance between two clusters?

There are several methods.

Single linkage

Take the closest pair.

$$ d(A,B)=\min d(x,y) $$
Complete linkage

Take the farthest pair.

$$ d(A,B)=\max d(x,y) $$
Average linkage

Take the average distance between all pairs.

Ward linkage

Merge clusters based on the increase in within-cluster variance.

We'll study each mathematically

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df[["Hours_Studied","Exam_Score"]])
model=AgglomerativeClustering(n_clusters=2
                              ,
                              linkage="ward"
                              )

labels=model.fit_predict(X_scaled)
print(labels)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering

# Select features
X = df[["Hours_Studied", "Exam_Score"]]

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Hierarchical clustering
model = AgglomerativeClustering(
    n_clusters=2,
    linkage="ward"
)

labels = model.fit_predict(X_scaled)

print(labels)
```

```python
df["Cluster"] = labels

print(df)
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    df["Hours_Studied"],
    df["Exam_Score"],
    c=df["Cluster"]
)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Hierarchical Clustering with Ward Linkage")

plt.show()
```
