```python
df_encoded = pd.get_dummies(
    df,
    columns=[
        "Gender",
        "Ever_Married",
        "Graduated",
        "Profession",
        "Spending_Score",
        "Var_1"
    ],
    dtype=int
)

df_encoded.head()
```

```python
print("Original shape:", df.shape)
print("Encoded shape:", df_encoded.shape)
```

```python
X=df_encoded
X.head()
```

```python
X.dtypes
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
print(X_scaled.shape)
```

```python
print("Original shape:", df.shape)
print("Encoded shape:", df_encoded.shape)
print(X_scaled.shape)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inertias = []
silhouette_scores = []

K_range = range(2, 11)

for k in K_range:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    inertias.append(model.inertia_)
    silhouette_scores.append(
        silhouette_score(X_scaled, labels)
    )

    print(
        f"K = {k} | "
        f"Inertia = {model.inertia_:.2f} | "
        f"Silhouette = {silhouette_scores[-1]:.3f}"
    )
```

```python
plt.figure(figsize=(10,5))
plt.plot(K_range, inertias,
         marker="o")
plt.xlabel("K")
plt.ylabel("Inertia")
plt.title("ElbOW METHOD")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(K_range, silhouette_scores, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score vs K")

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(X_scaled)

print(df["Cluster"].value_counts().sort_index())
```

```python
cluster_profile = df.groupby("Cluster").mean(numeric_only=True)

cluster_profile
```

```python
categorical_cols = [
    "Gender",
    "Ever_Married",
    "Graduated",
    "Profession",
    "Spending_Score",
    "Var_1"
]

for col in categorical_cols:
    print("\n" + "=" * 50)
    print(col)
    print(
        pd.crosstab(
            df["Cluster"],
            df[col],
            normalize="index"
        ).round(2)
    )
```

```python
# ============================================================
# CUSTOMER SEGMENTATION - COMPLETE VISUALIZATION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# Make sure df already contains the "Cluster" column
# ------------------------------------------------------------

sns.set_theme(style="whitegrid")

# ============================================================
# 1. CLUSTER SIZE - BAR CHART
# ============================================================

cluster_counts = df["Cluster"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.xlabel("Cluster")
plt.ylabel("Number of Customers")
plt.title("Number of Customers in Each Cluster")

for i, value in enumerate(cluster_counts.values):
    plt.text(i, value + 20, str(value), ha="center")

plt.show()


# ============================================================
# 2. CLUSTER DISTRIBUTION - PIE CHART
# ============================================================

plt.figure(figsize=(7, 7))

plt.pie(
    cluster_counts.values,
    labels=[f"Cluster {x}" for x in cluster_counts.index],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Distribution by Cluster")

plt.show()


# ============================================================
# 3. AGE vs FAMILY SIZE - SCATTER PLOT
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Family_Size",
    hue="Cluster",
    palette="tab10",
    s=70,
    alpha=0.7
)

plt.title("Age vs Family Size by Cluster")
plt.xlabel("Age")
plt.ylabel("Family Size")
plt.legend(title="Cluster")

plt.show()


# ============================================================
# 4. AGE vs WORK EXPERIENCE - SCATTER PLOT
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Work_Experience",
    hue="Cluster",
    palette="tab10",
    s=70,
    alpha=0.7
)

plt.title("Age vs Work Experience by Cluster")
plt.xlabel("Age")
plt.ylabel("Work Experience")
plt.legend(title="Cluster")

plt.show()


# ============================================================
# 5. AGE DISTRIBUTION - BOX PLOT
# ============================================================

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Cluster",
    y="Age"
)

plt.title("Age Distribution Across Clusters")
plt.xlabel("Cluster")
plt.ylabel("Age")

plt.show()


# ============================================================
# 6. WORK EXPERIENCE - BOX PLOT
# ============================================================

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Cluster",
    y="Work_Experience"
)

plt.title("Work Experience Across Clusters")
plt.xlabel("Cluster")
plt.ylabel("Work Experience")

plt.show()


# ============================================================
# 7. AGE DISTRIBUTION - VIOLIN PLOT
# ============================================================

plt.figure(figsize=(9, 6))

sns.violinplot(
    data=df,
    x="Cluster",
    y="Age"
)

plt.title("Age Distribution Across Clusters - Violin Plot")
plt.xlabel("Cluster")
plt.ylabel("Age")

plt.show()


# ============================================================
# 8. SPENDING SCORE BY CLUSTER
# ============================================================

spending_table = pd.crosstab(
    df["Cluster"],
    df["Spending_Score"],
    normalize="index"
) * 100

spending_table.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Spending Score Distribution by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Spending Score")

plt.show()


# ============================================================
# 9. PROFESSION BY CLUSTER
# ============================================================

profession_table = pd.crosstab(
    df["Cluster"],
    df["Profession"]
)

profession_table.plot(
    kind="bar",
    figsize=(13, 6)
)

plt.title("Profession Distribution by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.legend(title="Profession", bbox_to_anchor=(1.05, 1))

plt.tight_layout()
plt.show()


# ============================================================
# 10. NUMERIC CLUSTER PROFILE - HEATMAP
# ============================================================

numeric_features = [
    "Age",
    "Work_Experience",
    "Family_Size"
]

cluster_profile = df.groupby("Cluster")[numeric_features].mean()

plt.figure(figsize=(9, 5))

sns.heatmap(
    cluster_profile,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd"
)

plt.title("Average Numeric Features by Cluster")
plt.xlabel("Features")
plt.ylabel("Cluster")

plt.show()


# ============================================================
# 11. CLUSTER-WISE AVERAGE FEATURES - BAR CHART
# ============================================================

cluster_profile.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Numeric Features Across Clusters")
plt.xlabel("Cluster")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.legend(title="Feature")

plt.show()


# ============================================================
# 12. PCA 2D VISUALIZATION
# ============================================================

from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

pca_df["Cluster"] = df["Cluster"].values

plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Cluster",
    palette="tab10",
    s=80,
    alpha=0.75
)

plt.title("Customer Clusters in 2D PCA Space")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Cluster")

plt.show()


# ============================================================
# 13. CLUSTER vs GENDER
# ============================================================

gender_table = pd.crosstab(
    df["Cluster"],
    df["Gender"],
    normalize="index"
) * 100

gender_table.plot(
    kind="bar",
    figsize=(9, 6)
)

plt.title("Gender Distribution by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.legend(title="Gender")

plt.show()


# ============================================================
# 14. CLUSTER vs EDUCATION
# ============================================================

education_table = pd.crosstab(
    df["Cluster"],
    df["Graduated"],
    normalize="index"
) * 100

education_table.plot(
    kind="bar",
    figsize=(9, 6)
)

plt.title("Graduation Status by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.legend(title="Graduated")

plt.show()


# ============================================================
# 15. FINAL CLUSTER SUMMARY
# ============================================================

final_summary = df.groupby("Cluster").agg(
    Customers=("Cluster", "size"),
    Average_Age=("Age", "mean"),
    Average_Work_Experience=("Work_Experience", "mean"),
    Average_Family_Size=("Family_Size", "mean")
)

print("\n" + "=" * 60)
print("FINAL CUSTOMER SEGMENTATION SUMMARY")
print("=" * 60)

print(final_summary.round(2))
```
