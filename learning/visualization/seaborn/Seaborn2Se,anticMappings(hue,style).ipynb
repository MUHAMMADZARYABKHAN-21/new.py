```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Apply a professional theme
sns.set_theme(style="whitegrid")

# Load the Titanic dataset
df = pd.read_csv("Titanic1.csv")

# Remove rows with missing values in required columns
df = df.dropna(subset=["age", "fare", "sex", "survived"])

# Create the figure
plt.figure(figsize=(10, 6))

# Scatter plot
sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived",      # Color by survival
    style="sex",         # Marker shape by gender
    size="fare",         # Marker size by fare
    palette="Set1",      # Color palette
    sizes=(30, 300),     # Minimum and maximum marker size
    alpha=0.8            # Transparency
)

# Titles and labels
plt.title("Titanic Dataset: Age vs Fare", fontsize=16)
plt.xlabel("Age", fontsize=12)
plt.ylabel("Fare", fontsize=12)

# Display the plot
plt.show()
```
