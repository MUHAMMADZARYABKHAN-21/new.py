```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset directly from seaborn
df = sns.load_dataset("titanic")

# Select numeric columns
numeric_df = df[["survived", "pclass", "age", "sibsp", "parch", "fare"]]

# Create PairGrid
g = sns.PairGrid(numeric_df)
g.map(sns.scatterplot)

plt.show()

g = sns.PairGrid(numeric_df)

g.map_diag(sns.histplot)

plt.show()

g=sns.PairGrid(numeric_df)
g.map_diag(sns.histplot)
g.map_offdiag(sns.kdeplot)
plt.show()
```

# JointGrid

In this notebook, you will learn:

- What is JointGrid?
- Joint plot vs JointGrid
- Joint plot
- Marginal plots
- map()
- plot_joint()
- plot_marginals()
- Different plots for the center and margins
- hue
- Customizing JointGrid

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset("titanic")

g = sns.JointGrid(
    data=df,
    x="age",
    y="fare"
)

g.plot(
    sns.scatterplot,
    sns.histplot
)


plt.show()

g = sns.JointGrid(
    data=df,
    x="age",
    y="fare"
)

g.plot_joint(
    sns.scatterplot
)

g.plot_marginals(
    sns.histplot
)

plt.show()
```
