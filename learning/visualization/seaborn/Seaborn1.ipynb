```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("HotelReservation.csv")

adults = data["no_of_adults"]
children = data["no_of_children"]

plt.hist(adults, bins=10, alpha=0.7, label="Adults")
plt.hist(children, bins=10, alpha=0.7, label="Children")

plt.xlabel("Count")
plt.ylabel("Frequency")
plt.title("Distribution of Adults and Children")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load your dataset
data = pd.read_csv("HotelReservation.csv")

# Step 1: Check the actual column names
print("Columns in dataset:", data.columns)

# Step 2: Use the correct column name (example: 'room_type_reserved')
# Replace 'room_type_reserved' with whatever name you see in the print output
sns.countplot(data=data,
              x="room_type_reserved",
              order=data["room_type_reserved"].value_counts().index)

plt.title("Number of Reservations per Room Type")
plt.xlabel("Room Type")
plt.ylabel("Number of Reservations")
plt.show()
```
