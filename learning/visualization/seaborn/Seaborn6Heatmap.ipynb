- ## Correlation

Correlation measures how two numerical variables are related.

Correlation values range from:

- +1 → Strong positive relationship
-  0 → No linear relationship
- -1 → Strong negative relationship

Examples:

+0.90 → Very strong positive correlation
+0.50 → Moderate positive correlation
 0.00 → Little/no linear correlation
-0.50 → Moderate negative correlation
-0.90 → Very strong negative correlation

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pkg_resources import distributions_from_metadata

# Load Titanic dataset
df = pd.read_csv("Titanic1.csv")

# Compute correlation matrix for selected numeric columns
corr = df[["survived","pclass","age","sibsp","parch","fare"]].corr()

# Plot heatmap of correlations
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0)

plt.title("Correlation Heatmap (Titanic Features)")
plt.show()

```

In this notebook, you will learn:

🧠 regplot() vs lmplot()

This is important.

regplot()	lmplot()
Creates one regression plot	Can create multiple plots
Works naturally with Matplotlib axes	Built around Seaborn's FacetGrid
Good for simple regression	Excellent for grouped comparisons
More flexible for individual axes	Better for faceting


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv("Titanic1.csv")

df.head()

sns.lmplot(
    data=df,
    x="age",
    y="fare"
)
plt.show()

sns.lmplot(
    data=df,
    x="age",
    y="fare",
    hue="sex",
    col="pclass",


)

plt.show()


sns.lmplot(
    data=df,
    x="age",
    y="fare",
    col="pclass",
    hue="sex"
)
plt.show()



sns.lmplot(
    data=df,
    x="age",
    y="fare",
    row="pclass"
)

plt.show()





```

💻 Cell 8 — Scatter Plot with FacetGrid

FacetGrid isn't limited to histograms.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("Titanic1.csv")


g=sns.FacetGrid(
    df,
    col="pclass",
)
g.map(sns.scatterplot,"age","fare")
plt.show()

g=sns.FacetGrid(
    df,col="pclass",
    hue="sex"
)
g.map(sns.scatterplot,"age","fare")
g.add_legend()
plt.show()
```
