```python
import matplotlib.pyplot as plt
import pandas as pd

dates = pd.date_range("2025-01-01", periods=7)

sales = [120, 150, 180, 160, 190, 220, 210]

plt.figure(figsize=(8,5))

plt.plot(dates, sales, marker="o")

plt.title("Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.grid(True)
plt.xticks(rotation=90)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

dates = pd.date_range("1912-04-01", periods=10)

tickets = [50, 65, 80, 75, 90, 110, 95, 120, 140, 160]

plt.figure(figsize=(8,5))

plt.plot(
    dates,
    tickets,
    marker="o",
    color="green"
)

plt.title("Ticket Sales Before Departure")

plt.xlabel("Date")
plt.ylabel("Tickets Sold")

plt.xticks(rotation=45)

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

dates = pd.date_range("2025-01-01", periods=5)

sales = [10,20,25,30,40]
profit = [2,5,7,10,12]

plt.plot(
    dates,
    sales,
    marker="o",
    label="Sales"
)

plt.plot(
    dates,
    profit,
    marker="s",
    label="Profit"
)

plt.legend()

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample dates and values
dates = pd.date_range("2026-08-01", periods=7, freq="D")
values = [10, 15, 20, 18, 25, 22, 30]

plt.figure(figsize=(8,5))
plt.plot(dates, values, marker="o", color="blue")

plt.title("Daily Values")
plt.xlabel("Date")
plt.ylabel("Value")
plt.grid(True)
plt.show()
```

```python
dates = pd.date_range("2026-08-01", periods=7, freq="D")
sales_A = [100, 120, 115, 130, 125, 140, 150]
sales_B = [90, 95, 100, 110, 105, 115, 120]

plt.figure(figsize=(8,5))
plt.plot(dates, sales_A, marker="o", label="Product A", color="red")
plt.plot(dates, sales_B, marker="o", label="Product B", color="orange")

plt.title("Sales Comparison Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```
