```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [22, 11, 6, 44]

plt.plot(x, y, marker="o", linestyle="--", color="red", linewidth=2)

# Add text annotations
plt.text(1, 22, "Point 1: Highest", fontsize=12, color="red")
plt.text(2, 11, "Point 2: Lowest", fontsize=12, color="blue")
plt.text(3, 6, "Point 3", fontsize=12, color="green")
plt.text(4, 44, "Point 4", fontsize=12, color="purple")

plt.title("Annotated Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,30]

plt.plot(x, y, marker="o")

plt.annotate(
    "Highest Sales",
    xy=(4,30),
    xytext=(3,35),
    arrowprops=dict(arrowstyle="->")
)

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Ali", "Sara", "John"]
marks = [85, 92, 78]

plt.bar(students, marks)

for i in range(len(students)):
    plt.text(i, marks[i], str(marks[i]), ha="center")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = ["ALI", "SARA", "JOHN", "Noman"]
y = [67, 78, 99, 100]

plt.plot(x, y, marker="o", linestyle="--", color="red", linewidth=2)

# Annotate the highest sales point (Noman, 100)
plt.annotate(
    "Highest Sales",
    xy=("Noman", 100),          # point to annotate
    xytext=("JOHN", 90),        # text position
    arrowprops=dict(arrowstyle="->", color="blue")
)

plt.title("Sales Performance")
plt.xlabel("Name")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("titanic.csv")

count = df["sex"].value_counts()

plt.bar(count.index, count.values)

for i, value in enumerate(count.values):
    plt.text(i, value, str(value), ha="center")

plt.title("Passengers by Gender")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Create figure and axis
fig, ax = plt.subplots()
x = np.linspace(0, 2*np.pi, 200)
line, = ax.plot(x, np.sin(x), color="blue")

# Update function for animation
def update(frame):
    line.set_ydata(np.sin(x + frame / 10))  # shift sine wave
    return line,

# Create animation
ani = FuncAnimation(fig, update, frames=100, interval=100, blit=True)

plt.title("Animated Sine Wave")
plt.xlabel("X")
plt.ylabel("sin(X)")
plt.grid(True)
plt.show()
```
