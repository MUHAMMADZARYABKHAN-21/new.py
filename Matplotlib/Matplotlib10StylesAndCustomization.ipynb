```python
import matplotlib.pyplot as plt
import pandas as pd

# Read dataset
df = pd.read_csv("Titanic.csv")

# Summary
print(df.describe())
print(df.info())
print(df.isnull().sum())

# Fill missing values in age
age_mean = df["age"].mean()
df["age"] = df["age"].fillna(age_mean)

# Display columns
print(df["age"])
print(df[["age", "survived"]])

# Plot Age Distribution
plt.figure(figsize=(6, 5))

plt.hist(
    df["age"],
    bins=20,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Example Bitcoin price history (Date + Close)
data = {
    "Date": [
        "2026-07-25", "2026-07-26", "2026-07-27",
        "2026-07-28", "2026-07-29", "2026-07-30",
        "2026-07-31", "2026-08-01"
    ],
    "Close": [
        64311.81, 65340.30, 63724.90,
        63871.36, 63908.17, 64725.31,
        62813.75, 62763.32
    ]
}

# Convert to DataFrame
df = pd.DataFrame(data)
df["Date"] = pd.to_datetime(df["Date"])

# Plot Bitcoin prices
plt.figure(figsize=(10,5))
plt.plot(df["Date"], df["Close"], color="orange", linewidth=2, marker="o")

# Titles and labels
plt.title("Bitcoin Daily Closing Price (USD)", fontsize=16, fontweight="bold")
plt.xlabel("Date", fontsize=12)
plt.ylabel("Price (USD)", fontsize=12)

# Grid and formatting
plt.grid(True, linestyle="--", alpha=0.6)
plt.xticks(rotation=45)

# Show chart
plt.show()
```
