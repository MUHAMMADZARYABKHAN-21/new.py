```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(
    months,
    sales,
    marker="o",
    color="blue",
    linewidth=2,
    label="Sales"
)

ax.set_title("Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue")
ax.grid(True)
ax.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a Figure with 2 Axes (side by side)
fig, axs = plt.subplots(1, 2, figsize=(10,4))

# First Axes
axs[0].plot([1,2,3,4], [10,20,15,25], color="blue")
axs[0].set_title("Line Plot")

# Second Axes
axs[1].bar([1,2,3,4], [5,7,3,8], color="green")
axs[1].set_title("Bar Chart")

# Show the Figure
plt.suptitle("Figure with Two Axes")
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2,2, figsize=(8,6))

axes[0,0].plot([1,2,3],[2,4,6])
axes[0,0].set_title("Line")

axes[0,1].bar([1,2,3],[2,4,6])
axes[0,1].set_title("Bar")

axes[1,0].scatter([1,2,3],[2,4,6])
axes[1,0].set_title("Scatter")

axes[1,1].hist([1,2,2,3,3,3,4])
axes[1,1].set_title("Histogram")

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Create 2x2 subplots
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

# Line plot
axes[0, 0].plot([1, 2, 3], [2, 4, 6], color="blue", label="Line Plot")
axes[0, 0].set_title("Line")
axes[0, 0].legend()   # legend for this subplot

# Bar plot
axes[0, 1].bar([1, 2, 3], [2, 4, 6], label="Bar Plot")
axes[0, 1].set_title("Bar")
axes[0, 1].legend()

# Scatter plot
axes[1, 0].scatter([1, 2, 3], [2, 4, 6], label="Scatter Plot")
axes[1, 0].set_title("Scatter")
axes[1, 0].legend()

# Histogram
axes[1, 1].hist([1, 2, 2, 3, 3, 3, 4], label="Histogram")
axes[1, 1].set_title("Histogram")
axes[1, 1].legend()

# Adjust layout
plt.tight_layout()
plt.show()
```
