```python
import matplotlib.pyplot as plt

# Students
students = ["Ali", "Sara", "John", "Noman"]

# Marks in two subjects
math_marks = [67, 78, 99, 100]
science_marks = [70, 82, 95, 90]

# X-axis positions
x = range(len(students))

# Plot bars side by side
plt.bar([i - 0.2 for i in x], math_marks, width=0.4, label="Math", color="blue")
plt.bar([i + 0.2 for i in x], science_marks, width=0.4, label="Science", color="green")

# Labels and title
plt.xticks(x, students)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Comparison of Math and Science Marks")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("titanic.csv")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(10,4),
    sharey=True
)

axes[0].hist(df["age"].dropna(), bins=10)
axes[0].set_title("Age")

axes[1].hist(df["fare"].dropna(), bins=10)
axes[1].set_title("Fare")

plt.tight_layout()

plt.show()
```
