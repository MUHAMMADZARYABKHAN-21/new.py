```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [2, 4, 6, 8]

# Create a figure with 1 row, 2 columns of subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# First subplot: Line plot
axes[0].plot(x, y, marker="o", color="blue")
axes[0].set_title("Line Plot")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")

# Second subplot: Bar chart
axes[1].bar(x, y, color="green", edgecolor="black")
axes[1].set_title("Bar Chart")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Y")

plt.tight_layout()  # Adjust spacing so titles/labels don’t overlap
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

# Create figure with size
plt.figure(figsize=(10, 4))

# First subplot
plt.subplot(1, 2, 1)
plt.plot(x, y, marker="o", color="blue")
plt.title("Line Plot")
plt.xlabel("Samples")
plt.ylabel("Precision")

# Second subplot
plt.subplot(1, 2, 2)
plt.bar(x, y, color="green", edgecolor="black")
plt.title("Sales")
plt.xlabel("Samples")
plt.ylabel("Values")

# Save before showing
plt.savefig("sales.png")          # default PNG
plt.savefig("graph.png")          # another PNG
plt.savefig("sales.pdf")          # PDF format
plt.savefig("sales_highres.png",  # high resolution
            bbox_inches="tight",
            dpi=300)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a figure
plt.figure(figsize=(8, 6))

# ---------------- Subplot 1: Bar Chart ----------------
plt.subplot(2, 2, 1)
plt.bar(
    [1, 2, 3, 4],
    [33, 22, 11, 11],
    color="red",
    edgecolor="black"
)
plt.title("Bar Chart")
plt.xlabel("Category")
plt.ylabel("Value")

# ---------------- Subplot 2: Pie Chart ----------------
plt.subplot(2, 2, 2)
plt.pie(
    [33, 22, 11, 11],
    labels=["A", "B", "C", "D"],
    colors=["blue", "green", "orange", "purple"],
    autopct="%1.1f%%"
)
plt.title("Pie Chart")

# ---------------- Subplot 3: Line Plot ----------------
plt.subplot(2, 2, 3)
plt.plot(
    [1, 2, 3, 4],
    [33, 22, 11, 11],
    color="red",
    marker="o",
    linewidth=2
)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)

# ---------------- Subplot 4: Histogram ----------------
plt.subplot(2, 2, 4)
plt.hist(
    [1, 2, 2, 3, 3, 3, 4],
    bins=4,
    color="skyblue",
    edgecolor="black"
)
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

# Adjust spacing
plt.tight_layout()

# Display all plots
plt.show()
```
