```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

x = [1,2,3,4,5]
y = [2,4,5,4,5]
z = [5,6,7,8,9]

fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(x, y, z)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Scatter Plot")

plt.show()
```

```python
import numpy as np

x = np.linspace(-5,5,50)
y = np.linspace(-5,5,50)

X, Y = np.meshgrid(x,y)
Z = np.sin(np.sqrt(X**2 + Y**2))
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(
    X,
    Y,
    Z,
    cmap="viridis"
)
ax.plot_wireframe(
    X,
    Y,
    Z
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("titanic.csv")

fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    df["age"],
    df["fare"],
    df["pclass"]
)

ax.set_xlabel("Age")
ax.set_ylabel("Fare")
ax.set_zlabel("Passenger Class")

plt.show()
```
